## Passo 4 — Parâmetros da busca e custo estimado

A query usa os [operadores de busca do X](https://docs.x.com/x-api/posts/search/integrate/build-a-query). Os filtros reduzem o custo, porque só posts devolvidos são cobrados:

* `lang:pt`: só português.
* `-is:retweet`: retweets são duplicatas (na Atividade 1.3 eles eram descartados na Silver; aqui nem chegam a ser pagos).
* `has:geo`: só posts **com localização** (ponto exato ou local marcado). Sem localização não há latitude/longitude nem município, então nem vale pagar por eles.
* Palavras de evento **E** cidades dos CDs da LODLog.

O X não tem operador para "só ponto exato" nem para "só local do tamanho de uma cidade": o `has:geo` também aceita posts marcados com um estado ou o país inteiro. Esse segundo filtro fica na **Silver** (Passo 7), e o município é confirmado pela coordenada no **Passo 9**. Posts descartados ali já foram pagos.

Com `has:geo` o volume cai muito, porque a maioria dos posts não tem localização. Se a coleta vier vazia, aumente `MAX_PAGINAS` ou tire uma das restrições de texto; o custo continua limitado aos posts devolvidos.

A API devolve de 10 a 100 posts por página. Comece com o padrão e só aumente se precisar.

````python
QUERY = (
    "(bloqueio OR bloqueada OR greve OR protesto OR acidente OR tombamento "
    "OR congestionamento OR interditada) "
    '(Campinas OR "São Paulo" OR "Belo Horizonte" OR Curitiba OR "Porto Alegre") '
    "lang:pt -is:retweet has:geo"
)
MAX_RESULTS = 10      # posts por página (10 a 100)
MAX_PAGINAS = 1       # páginas no máximo
PRECO_POR_POST = 0.005  # US$ — confira no Developer Console

print(QUERY)
print(f"Custo máximo estimado: {MAX_RESULTS * MAX_PAGINAS} posts × US$ {PRECO_POR_POST} "
      f"= US$ {MAX_RESULTS * MAX_PAGINAS * PRECO_POR_POST:.2f}")
````

## Passo 5 — Coleta (chamada à API)

Cada página da API traz uma lista de posts em `data` e, separadamente, os usuários e locais referenciados em `includes`. Para manter **o mesmo formato da Bronze da Atividade 1.3**, a função `desmontar_pagina` gera **um registro por post**, cada um com o seu `includes`.

**Localização.** A API só devolve onde o post foi feito se a gente pedir. Há duas fontes, e as duas são opcionais (a maioria dos posts não tem nenhuma):

* `data.geo.coordinates`: **ponto exato** (`[longitude, latitude]`), quando o autor marcou a localização precisa. É raro.
* `includes.places[].geo.bbox`: **retângulo** do local marcado (`[lon_oeste, lat_sul, lon_leste, lat_norte]`), que pode ser um bairro, uma cidade, um estado ou o país inteiro (`place_type`).

Por isso pedimos `geo` em `tweet.fields`, a expansão `geo.place_id` e `geo,place_type` em `place.fields`.

O arquivo é gravado **exatamente como veio da API** (fora essa separação por post). Nada é filtrado ou renomeado aqui. Essa é a zona de pouso.

````python
import json
import time
from datetime import datetime, timezone
from pathlib import Path

URL_BUSCA = "https://api.x.com/2/tweets/search/recent"
CAMPOS = {
    "tweet.fields": "created_at,author_id,lang,public_metrics,entities,geo,referenced_tweets",
    "expansions": "author_id,geo.place_id",
    "user.fields": "username",
    "place.fields": "full_name,place_type,country_code,geo",
}

def desmontar_pagina(resposta):
    includes = resposta.get("includes", {})
    usuarios = {u["id"]: u for u in includes.get("users", [])}
    locais = {p["id"]: p for p in includes.get("places", [])}
    registros = []
    for post in resposta.get("data", []):
        inc = {}
        if post.get("author_id") in usuarios:
            inc["users"] = [usuarios[post["author_id"]]]
        place_id = post.get("geo", {}).get("place_id")
        if place_id in locais:
            inc["places"] = [locais[place_id]]
        registros.append({"data": post, "includes": inc})
    return registros

def coletar(query, max_results, max_paginas, token):
    headers = {"Authorization": f"Bearer {token}"}
    params = {**CAMPOS, "query": query, "max_results": max_results}
    registros, paginas = [], 0
    while paginas < max_paginas:
        r = requests.get(URL_BUSCA, headers=headers, params=params, timeout=30)
        if r.status_code == 429:
            reset = int(r.headers.get("x-rate-limit-reset", time.time() + 60))
            espera = min(max(reset - time.time(), 1), 900)
            print(f"Limite de requisições atingido; aguardando {espera:.0f}s...")
            time.sleep(espera)
            continue
        if r.status_code != 200:
            raise RuntimeError(f"HTTP {r.status_code}: {r.text[:500]}")
        resposta = r.json()
        paginas += 1
        registros += desmontar_pagina(resposta)
        meta = resposta.get("meta", {})
        print(f"Página {paginas}: {meta.get('result_count', 0)} posts")
        if "next_token" not in meta:
            break
        params["next_token"] = meta["next_token"]
    return registros

registros = coletar(QUERY, MAX_RESULTS, MAX_PAGINAS, X_BEARER_TOKEN)
print(f"Total coletado: {len(registros)} posts "
      f"(custo aproximado: US$ {len(registros) * PRECO_POR_POST:.2f})")
````

````python
# Zona de pouso: Volume no Databricks; pasta local no Plano B
if EM_DATABRICKS:
    spark.sql("CREATE SCHEMA IF NOT EXISTS lodlog_lake")
    spark.sql("CREATE VOLUME IF NOT EXISTS lodlog_lake.raw")
    catalogo = spark.catalog.currentCatalog()
    PASTA_POUSO = Path(f"/Volumes/{catalogo}/lodlog_lake/raw/x_posts")
else:
    PASTA_POUSO = Path("x_posts")
PASTA_POUSO.mkdir(parents=True, exist_ok=True)

carimbo = datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%SZ")
arquivo = PASTA_POUSO / f"x_posts_{carimbo}.jsonl"
with open(arquivo, "w", encoding="utf-8") as f:
    for reg in registros:
        f.write(json.dumps(reg, ensure_ascii=False) + "\n")
print(f"{len(registros)} posts gravados em {arquivo}")
````

## Passo 6 — Bronze (no Databricks)

**Plano B:** se a coleta rodou fora do Databricks, crie o Volume com a célula SQL abaixo e faça upload dos arquivos `x_posts_*.jsonl`. No menu **Catalog**, abra `workspace` → `lodlog_lake` → `raw`, crie a pasta `x_posts` e use **Upload to this volume**.

````python
%sql
CREATE SCHEMA IF NOT EXISTS lodlog_lake;
CREATE VOLUME IF NOT EXISTS lodlog_lake.raw;

CREATE TABLE IF NOT EXISTS lodlog_lake.x_posts_bronze_api (
  post_id        STRING,
  payload        STRING,
  arquivo_origem STRING,
  ingerido_em    TIMESTAMP
) USING DELTA;
````

A carga lê **todos** os arquivos da zona de pouso e usa `MERGE` por `post_id`: posts já carregados são ignorados, só os novos entram. A carga é **idempotente**, ou seja, pode ser repetida sem duplicar dados. O mesmo post também pode vir em duas coletas seguidas, e a deduplicação na leitura (`dropDuplicates`) cuida disso.

````python
from pyspark.sql import functions as F

catalogo = spark.catalog.currentCatalog()
novos = (spark.read.text(f"/Volumes/{catalogo}/lodlog_lake/raw/x_posts/*.jsonl")
         .select(F.get_json_object("value", "$.data.id").alias("post_id"),
                 F.col("value").alias("payload"),
                 F.col("_metadata.file_name").alias("arquivo_origem"),
                 F.current_timestamp().alias("ingerido_em"))
         .where("post_id IS NOT NULL")
         .dropDuplicates(["post_id"]))
novos.createOrReplaceTempView("x_posts_novos")

spark.sql('''
MERGE INTO lodlog_lake.x_posts_bronze_api b
USING x_posts_novos n
ON b.post_id = n.post_id
WHEN NOT MATCHED THEN INSERT *
''').show()

spark.sql("SELECT COUNT(*) AS posts_na_bronze FROM lodlog_lake.x_posts_bronze_api").show()
````

## Passo 7 — Silver: mesma regra da Atividade 1.3

A regra de classificação é **idêntica** à da Atividade 1.3, para vocês compararem como ela se sai com dados reais.

A Silver ganha `latitude`, `longitude` e `origem_coord`:

* `ponto exato`: veio de `data.geo.coordinates`.
* `centro do local`: centro do retângulo (`bbox`) do local marcado, **só** quando o local é pequeno (`city`, `neighborhood` ou `poi`). É uma aproximação: o centro do retângulo nem sempre cai dentro do próprio local. O centro do retângulo de um estado ou do Brasil inteiro cairia num município qualquer, e isso seria um erro, não uma localização.
* sem coordenada (local marcado grande demais, como estado ou país): o post é **descartado**. Quotes e respostas (`quoted`, `replied_to`) são mantidos, e o autor **não** passa para a Silver (minimização, LGPD).

````python
%sql
CREATE OR REPLACE TABLE lodlog_lake.x_eventos_silver_api USING DELTA AS
WITH posts AS (
  SELECT post_id,
         to_date(from_utc_timestamp(
           CAST(get_json_object(payload, '$.data.created_at') AS TIMESTAMP),
           'America/Sao_Paulo')) AS data_evento,
         lower(get_json_object(payload, '$.data.text')) AS texto,
         split(get_json_object(payload, '$.includes.places[0].full_name'), ',')[0] AS cidade_place,
         get_json_object(payload, '$.data.referenced_tweets[0].type') AS referencia,
         -- ponto exato: GeoJSON Point [longitude, latitude]
         CAST(get_json_object(payload, '$.data.geo.coordinates.coordinates[0]') AS DOUBLE) AS lon_ponto,
         CAST(get_json_object(payload, '$.data.geo.coordinates.coordinates[1]') AS DOUBLE) AS lat_ponto,
         -- local marcado: bbox [lon_oeste, lat_sul, lon_leste, lat_norte]
         get_json_object(payload, '$.includes.places[0].place_type') AS tipo_local,
         CAST(get_json_object(payload, '$.includes.places[0].geo.bbox[0]') AS DOUBLE) AS bbox_oeste,
         CAST(get_json_object(payload, '$.includes.places[0].geo.bbox[1]') AS DOUBLE) AS bbox_sul,
         CAST(get_json_object(payload, '$.includes.places[0].geo.bbox[2]') AS DOUBLE) AS bbox_leste,
         CAST(get_json_object(payload, '$.includes.places[0].geo.bbox[3]') AS DOUBLE) AS bbox_norte
  FROM lodlog_lake.x_posts_bronze_api
),
coords AS (
  SELECT *,
         tipo_local IN ('city', 'neighborhood', 'poi') AND bbox_oeste IS NOT NULL AS local_pequeno
  FROM posts
)
SELECT post_id,
       data_evento,
       COALESCE(cidade_place,
                CASE WHEN texto LIKE '%campinas%'       THEN 'Campinas'
                     WHEN texto LIKE '%são paulo%'      THEN 'São Paulo'
                     WHEN texto LIKE '%belo horizonte%' THEN 'Belo Horizonte'
                     WHEN texto LIKE '%curitiba%'       THEN 'Curitiba'
                     WHEN texto LIKE '%porto alegre%'   THEN 'Porto Alegre'
                END) AS cidade,
       CASE WHEN texto RLIKE 'bloque|protesto|manifesta'       THEN 'Bloqueio/Protesto'
            WHEN texto RLIKE 'greve|paralisa'                  THEN 'Greve'
            WHEN texto RLIKE 'acidente|tombamento|colis'       THEN 'Acidente'
            WHEN texto RLIKE 'congestionamento|lentid|parada'  THEN 'Congestionamento'
            ELSE 'Irrelevante'
       END AS tipo_evento,
       COALESCE(lat_ponto, CASE WHEN local_pequeno THEN (bbox_sul + bbox_norte) / 2 END)   AS latitude,
       COALESCE(lon_ponto, CASE WHEN local_pequeno THEN (bbox_oeste + bbox_leste) / 2 END) AS longitude,
       CASE WHEN lat_ponto IS NOT NULL THEN 'ponto exato'
            WHEN local_pequeno         THEN 'centro do local'
       END AS origem_coord
FROM coords
WHERE (referencia IS NULL OR referencia <> 'retweeted')
  AND (lat_ponto IS NOT NULL OR local_pequeno);  -- só posts com latitude/longitude

SELECT tipo_evento, cidade, origem_coord, COUNT(*) AS posts
FROM lodlog_lake.x_eventos_silver_api
GROUP BY tipo_evento, cidade, origem_coord
ORDER BY posts DESC;
````

## Passo 8 — Carregar o shapefile de municípios como tabela

Latitude e longitude sozinhas não dizem em que município o post foi feito. Para isso usamos a **malha municipal do IBGE** (`BR_Municipios_2025`), um *shapefile* com o polígono de cada um dos 5.573 municípios. Ele está no repositório em `Labs/shapes/BR_Municipios_2025_simplificado.zip`, com as fronteiras simplificadas a ≈55 m para caber no GitHub (detalhes em `Labs/shapes/README.md`).

Um shapefile não é uma tabela: é um conjunto de arquivos (`.shp` com as geometrias, `.dbf` com os atributos, `.prj` com o sistema de coordenadas). Vamos transformá-lo na tabela Delta `lodlog_lake.municipios_br`:

* atributos do IBGE: código (`cd_mun`), nome, UF, região e área;
* o polígono em texto **WKT** (*Well-Known Text*, ex.: `POLYGON ((-47.1 -22.9, ...))`), formato que qualquer banco ou biblioteca geoespacial lê;
* o **retângulo envolvente** (`lon_min`, `lat_min`, `lon_max`, `lat_max`), que permite um filtro rápido em SQL no Passo 9.

O CRS da malha é SIRGAS 2000 (EPSG:4674), em graus. A diferença para o WGS 84 dos posts do X é menor que 1 m, então usamos as coordenadas direto.

A célula abaixo instala `geopandas`, que lê shapefiles. O `%pip` reinicia o Python do notebook: as variáveis dos passos anteriores se perdem, mas as tabelas continuam lá.

````python
%pip install geopandas --quiet
````

````python
from pathlib import Path

import geopandas as gpd
import shapely

ARQUIVO_SHAPE = Path("shapes/BR_Municipios_2025_simplificado.zip").resolve()  # relativo à pasta Labs
TABELA = "lodlog_lake.municipios_br"

spark.sql("CREATE SCHEMA IF NOT EXISTS lodlog_lake")
if spark.catalog.tableExists(TABELA):
    print(f"{TABELA} já existe; nada a fazer. Para recarregar: DROP TABLE {TABELA}")
else:
    mun = gpd.read_file(ARQUIVO_SHAPE)
    print(f"{len(mun)} municípios lidos; CRS {mun.crs}")
    limites = mun.geometry.bounds  # minx, miny, maxx, maxy
    tabela = mun[["CD_MUN", "NM_MUN", "SIGLA_UF", "NM_REGIAO", "AREA_KM2"]].rename(columns=str.lower).assign(
        lon_min=limites.minx, lat_min=limites.miny, lon_max=limites.maxx, lat_max=limites.maxy,
        geom_wkt=shapely.to_wkt(mun.geometry.values, rounding_precision=6),
    )
    # Grava em lotes: os polígonos somam dezenas de MB de texto
    LOTE = 500
    for i in range(0, len(tabela), LOTE):
        modo = "overwrite" if i == 0 else "append"
        spark.createDataFrame(tabela.iloc[i:i + LOTE]).write.mode(modo).saveAsTable(TABELA)
    print(f"{TABELA}: {spark.table(TABELA).count()} linhas")
````

````python
%sql
SELECT cd_mun, nm_mun, sigla_uf, nm_regiao, area_km2,
       lon_min, lat_min, lon_max, lat_max,
       left(geom_wkt, 60) AS inicio_do_poligono
FROM lodlog_lake.municipios_br
WHERE nm_mun IN ('Campinas', 'São Paulo', 'Belo Horizonte', 'Curitiba', 'Porto Alegre')
ORDER BY nm_mun;
````

## Passo 9 — Geolocalizar os posts no município

A pergunta "em que município cai este ponto?" é um teste de **ponto-em-polígono**. Testar cada post contra os 5.573 polígonos seria caro, então fazemos em duas etapas:

1. **Filtro grosso em SQL:** o ponto precisa estar dentro do retângulo envolvente do município (`BETWEEN`). Sobram poucos candidatos por post, quase sempre 1 a 4.
2. **Filtro fino com a geometria:** só para os candidatos, o `shapely` testa se o ponto está dentro do polígono de verdade.

É o mesmo padrão dos índices espaciais dos bancos geográficos: descartar rápido o que com certeza não serve e calcular com precisão só o resto.

````python
import shapely
from pyspark.sql import functions as F

candidatos = spark.sql('''
SELECT s.post_id, s.latitude, s.longitude, s.origem_coord,
       m.cd_mun, m.nm_mun, m.sigla_uf, m.geom_wkt
FROM lodlog_lake.x_eventos_silver_api s
JOIN lodlog_lake.municipios_br m
  ON  s.longitude BETWEEN m.lon_min AND m.lon_max
  AND s.latitude  BETWEEN m.lat_min AND m.lat_max
WHERE s.latitude IS NOT NULL
''').toPandas()

if candidatos.empty:
    dentro = candidatos.drop(columns="geom_wkt")
else:
    poligonos = shapely.from_wkt(candidatos["geom_wkt"])
    contem = shapely.contains_xy(poligonos, candidatos["longitude"], candidatos["latitude"])
    dentro = candidatos.loc[contem].drop(columns="geom_wkt").drop_duplicates("post_id")

print(f"{candidatos['post_id'].nunique()} posts com coordenada; {len(candidatos)} pares post × município candidato; "
      f"{len(dentro)} posts geolocalizados")

esquema = "post_id STRING, latitude DOUBLE, longitude DOUBLE, origem_coord STRING, cd_mun STRING, municipio STRING, uf STRING"
geo = spark.createDataFrame(dentro.rename(columns={"nm_mun": "municipio", "sigla_uf": "uf"}), esquema)
geo.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("lodlog_lake.x_posts_municipio_api")
````

````python
%sql
-- Compare a cidade do texto/local marcado com o município obtido pela coordenada
SELECT s.post_id, s.tipo_evento, s.cidade AS cidade_texto_ou_local,
       g.municipio AS municipio_pela_coordenada, g.uf, s.origem_coord,
       ROUND(s.latitude, 4) AS lat, ROUND(s.longitude, 4) AS lon
FROM lodlog_lake.x_eventos_silver_api s
LEFT JOIN lodlog_lake.x_posts_municipio_api g ON g.post_id = s.post_id
ORDER BY g.municipio IS NULL, s.data_evento;
````

### Mapa: de onde vieram os posts

O mapa desenha a malha do IBGE e marca um ponto no **centroide** de cada município onde houve post geolocalizado no Passo 9. O tamanho do ponto é proporcional ao número de posts. O painel da esquerda mostra o Brasil inteiro, e o da direita aproxima a região dos pontos.

O centroide é calculado numa projeção métrica (SIRGAS 2000 / Brazil Polyconic, EPSG:5880), porque o centroide calculado direto em graus fica deslocado. O ponto representa o **município**, não o local exato do post: dois posts em bairros diferentes de São Paulo caem no mesmo ponto. É de propósito, para não expor a coordenada de quem postou (minimização, LGPD).

````python
from pathlib import Path

import geopandas as gpd
import matplotlib.pyplot as plt

ARQUIVO_SHAPE = Path("shapes/BR_Municipios_2025_simplificado.zip").resolve()  # relativo à pasta Labs
SUPERFICIE, MALHA, DIVISA_MUN, DIVISA_UF, PONTO, TEXTO = "#fcfcfb", "#e6e5e0", "#d6d5cf", "#a3a29b", "#2a78d6", "#52514e"

# Posts por município (saída do Passo 9)
posts_mun = spark.sql('''
SELECT cd_mun, municipio, uf, COUNT(*) AS posts
FROM lodlog_lake.x_posts_municipio_api
GROUP BY cd_mun, municipio, uf
ORDER BY posts DESC
''').toPandas()
display(posts_mun)  # a mesma informação do mapa, em tabela

mun = gpd.read_file(ARQUIVO_SHAPE)
# Divisas estaduais: a malha simplificada deixa frestas entre municípios vizinhos;
# um buffer de 100 m (em metros, EPSG:5880) fecha as frestas antes de juntar por UF
ufs = mun.to_crs(5880).assign(geometry=lambda g: g.buffer(100)).dissolve(by="SIGLA_UF").to_crs(mun.crs)
centroides = (mun.to_crs(5880).set_geometry(mun.to_crs(5880).centroid).to_crs(mun.crs)
              [["CD_MUN", "geometry"]].merge(posts_mun, left_on="CD_MUN", right_on="cd_mun"))

if centroides.empty:
    print("Nenhum post geolocalizado em município. Rode os Passos 5 a 9 com mais posts.")
else:
    tamanho = 40 + 260 * centroides["posts"] / centroides["posts"].max()
    fig, (ax_br, ax_zoom) = plt.subplots(1, 2, figsize=(14, 7), facecolor=SUPERFICIE,
                                         gridspec_kw={"width_ratios": [1, 1.2]})
    for ax in (ax_br, ax_zoom):
        ax.set_facecolor(SUPERFICIE)
        # No Brasil inteiro, as divisas municipais viram ruído; só o zoom as mostra
        mun.plot(ax=ax, color=MALHA, edgecolor=DIVISA_MUN if ax is ax_zoom else "none", linewidth=0.2)
        ufs.boundary.plot(ax=ax, color=DIVISA_UF, linewidth=0.5)
        centroides.plot(ax=ax, markersize=tamanho, color=PONTO, edgecolor=SUPERFICIE, linewidth=1, zorder=3)
        ax.set_axis_off()

    # Zoom: retângulo dos pontos com margem de 1,5° (no mínimo 4° de lado)
    x0, y0, x1, y1 = centroides.total_bounds
    cx, cy = (x0 + x1) / 2, (y0 + y1) / 2
    meio = max(x1 - x0, y1 - y0, 4) / 2 + 1.5
    ax_zoom.set_xlim(cx - meio, cx + meio)
    ax_zoom.set_ylim(cy - meio, cy + meio)

    # Rótulo direto só nos 8 municípios com mais posts, pulando vizinhos colados
    # a um ponto já rotulado (ex.: Guarulhos ao lado de São Paulo); a tabela acima traz todos
    rotulados = []
    for _, linha in centroides.nlargest(8, "posts").iterrows():
        x, y = linha.geometry.x, linha.geometry.y
        if any(abs(x - rx) < 0.08 * meio and abs(y - ry) < 0.08 * meio for rx, ry in rotulados):
            continue
        rotulados.append((x, y))
        ax_zoom.annotate(f"{linha['municipio']}/{linha['uf']} · {linha['posts']}",
                         (linha.geometry.x, linha.geometry.y), xytext=(8, 6),
                         textcoords="offset points", fontsize=9, color=TEXTO)

    # Legenda de tamanho (série única: a legenda explica o tamanho, não a cor)
    referencias = sorted({1, int(centroides["posts"].max())})
    for n in referencias:
        ax_br.scatter([], [], s=40 + 260 * n / centroides["posts"].max(), color=PONTO,
                        edgecolor=SUPERFICIE, label=f"{n} post{'s' if n > 1 else ''}")
    ax_br.legend(loc="lower left", frameon=False, labelcolor=TEXTO, title="Posts no município",
                   title_fontsize=9, fontsize=9, scatterpoints=1, borderpad=1.2, labelspacing=1.2)

    ax_br.set_title("Brasil", loc="left", fontsize=11, color=TEXTO)
    ax_zoom.set_title("Região dos posts", loc="left", fontsize=11, color=TEXTO)
    fig.suptitle(f"Posts do X geolocalizados por município: {int(centroides['posts'].sum())} posts "
                 f"em {len(centroides)} municípios", x=0.02, ha="left", fontsize=13, color="#0b0b0b")
    fig.text(0.02, 0.02, "Ponto no centroide do município (IBGE, malha 2025). Fonte: X API, busca recente (7 dias).",
             fontsize=8, color=TEXTO)
    plt.tight_layout(rect=(0, 0.04, 1, 0.95))
    plt.show()
````

## Passo 10 — Cruzar com as entregas da LODLog

As entregas do seed vão só até **30/09/2026**, e a busca recente cobre os **últimos 7 dias**. Por isso `entregas_no_dia` tende a sair **0** para os posts reais. A consulta ainda mostra, para cada evento, qual CD seria afetado e a taxa de atraso normal dele. Em produção, os dois lados seriam atuais.

Só entram posts **geolocalizados num município** no Passo 9 (`JOIN`, não `LEFT JOIN`). A cidade do evento segue a confiança de cada fonte:

* com **ponto exato**: o município da coordenada (Passo 9);
* sem ponto exato: o nome do local marcado ou, se não houver, o município do centro do local.

````python
%sql
WITH eventos_cidade AS (
  SELECT s.data_evento, s.tipo_evento, CASE WHEN g.origem_coord = 'ponto exato' THEN g.municipio
              ELSE COALESCE(s.cidade, g.municipio) END AS cidade
  FROM lodlog_lake.x_eventos_silver_api s
  JOIN lodlog_lake.x_posts_municipio_api g ON g.post_id = s.post_id
),
eventos AS (
  SELECT data_evento, cidade, concat_ws(', ', collect_set(tipo_evento)) AS eventos,
         COUNT(*) AS posts
  FROM eventos_cidade
  WHERE tipo_evento <> 'Irrelevante' AND cidade IS NOT NULL
  GROUP BY data_evento, cidade
),
normal_cd AS (
  SELECT sk_cd_origem, ROUND(100 * AVG(indicador_atraso), 1) AS pct_atraso_normal
  FROM lodlog_dw.fato_entregas
  GROUP BY sk_cd_origem
)
SELECT e.data_evento, e.cidade, e.eventos, e.posts, cd.codigo AS cd,
       COUNT(fe.sk_entrega)                    AS entregas_no_dia,
       ROUND(100 * AVG(fe.indicador_atraso), 1) AS pct_atraso_no_dia,
       n.pct_atraso_normal
FROM eventos e
JOIN lodlog_dw.dim_centro_distribuicao cd ON cd.municipio = e.cidade
JOIN normal_cd n                          ON n.sk_cd_origem = cd.sk_cd
LEFT JOIN lodlog_dw.fato_entregas fe      ON fe.sk_cd_origem = cd.sk_cd
                                         AND fe.data_entrega = e.data_evento
GROUP BY e.data_evento, e.cidade, e.eventos, e.posts, cd.codigo, n.pct_atraso_normal
ORDER BY e.data_evento;
````
